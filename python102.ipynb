## Classes

```python
class Founder:
    def __init__(self, name, company):
        self.name = name
        self.company = company

    def introduce(self):
        return f"I am {self.name}, the founder of {self.company}."
```

```python
founder= Founder("Abdirazak", "iKreatify")
```

```python
founder.introduce() 
```

```python
# class inheritance and polymorphism can be added here later

class CEO(Founder): 
    def __init__(self, name, company, years_experience):
        super().__init__(name, company)
        self.years_experience = years_experience

    def introduce(self):
        base_introduction = super().introduce()
        return f"{base_introduction} I have {self.years_experience} years of experience as a CEO."
```

```python
new_ceo = CEO("Satya Nadella", "Microsoft", 25)
new_ceo.introduce()
```

## Files and Error Handling

```python
# Write out to a file
with open("sample.txt", "w") as file:
    file.write("Hello, this is a sample file.\n")
    file.write("We are learning about file handling in Python.\n")
    file.write("This is the third line of the file.\n")

# Read from a file
with open("sample.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
# implement file error handling
try:
    with open("non_existent_file.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Error: The file does not exist.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
# implement file error handling with catching specific exceptions
try:
    with open("another_non_existent_file.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Error: The file was not found.")
except PermissionError:
    print("Error: You do not have permission to access this file.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
