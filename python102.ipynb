```python
names = ["Tim Cook", "Elon Musk", "Jeff Bezos"]
companies = ["Apple", "Tesla", "Amazon"]
```

```python
ceo_company_dict = dict(zip(names, companies))
```

```python
ceo_company_dict
```

```python
for ceo in ceo_company_dict:
    print(f"Hey {ceo}, you are the CEO of {ceo_company_dict[ceo]}!") 
```

```python
while ceo_company_dict:
    ceo, company = ceo_company_dict.popitem() # 
    print(f"{ceo} has left {company}.")
    
```

## Classes

```python
class Founder:
    def __init__(self, name, company):
        self.name = name
        self.company = company

    def introduce(self):
        return f"I am {self.name}, the founder of {self.company}."
```

```python
founder= Founder("Abdirazak", "iKreatify")
```

```python
founder.introduce() 
```

```python
# class inheritance and polymorphism can be added here later

class CEO(Founder): 
    def __init__(self, name, company, years_experience):
        super().__init__(name, company)
        self.years_experience = years_experience

    def introduce(self):
        base_introduction = super().introduce()
        return f"{base_introduction} I have {self.years_experience} years of experience as a CEO."
```

```python
new_ceo = CEO("Satya Nadella", "Microsoft", 25)
new_ceo.introduce()
```
